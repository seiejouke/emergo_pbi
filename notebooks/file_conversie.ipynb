```python
import pyodbc
import pandas as pd

mdb_path = r"/mnt/data/DemoDataBetsyBike.mdb"

# ODBC-connection string
conn_str = (
    r"DRIVER={Microsoft Access Driver (*.mdb, *.accdb)};"
    rf"DBQ={mdb_path};"
)

conn = pyodbc.connect(conn_str)
cursor = conn.cursor()

# alle tabellen ophalen
tables = [t.table_name for t in cursor.tables(tableType="TABLE")]

print("Gevonden tabellen:", tables)

for table in tables:
    df = pd.read_sql(f"SELECT * FROM [{table}]", conn)
    df.to_csv(f"{table}.csv", index=False)
    print(f"✔ {table}.csv opgeslagen")

conn.close()
```
